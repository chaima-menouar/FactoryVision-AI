```python
from pathlib import Path
DATA_ROOT = Path('/kaggle/working/datasets/MVTecAD')
OUTPUT_ROOT = Path('/kaggle/working/factoryvision-artifacts')
CATEGORIES = ['bottle', 'cable', 'metal_nut', 'transistor', 'zipper']
DATA_ROOT, OUTPUT_ROOT
```

```python
import json, pandas as pd
rows = []
for category in CATEGORIES:
    path = OUTPUT_ROOT / category / 'metrics.json'
    if path.exists():
        payload = json.loads(path.read_text())
        row = payload[0] if isinstance(payload, list) and payload else payload
        rows.append({'category': category, **row})
results = pd.DataFrame(rows)
display(results)
results.to_csv(OUTPUT_ROOT / 'baseline_summary.csv', index=False)
```
